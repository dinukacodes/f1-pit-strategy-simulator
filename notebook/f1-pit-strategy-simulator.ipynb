# F1 Pit-Stop Strategy Simulator: Safety-Car Uncertainty at Singapore

**Summary.** This notebook simulates a 62-lap Formula 1 race at Singapore to compare one-stop, two-stop and *pit-if-neutralised* tyre strategies when Safety Car (SC) and Virtual Safety Car (VSC) periods occur at random. Tyre wear follows a fuel-coupled, self-accelerating ODE solved with **RK4**. Track status follows a lap-wise **Markov chain** with a phase-type minimum Safety-Car length. Pit losses are derived from pit-lane geometry. Stationary crew times and lap noise are generated with **inverse-transform and Box–Muller** samplers. Strategies are compared with **two-stage Monte Carlo using common random numbers (CRN)**, reporting 95% confidence intervals for paired race-time differences. The simulator is verified against closed-form ODE solutions, a deterministic benchmark with a derived optimum, the Markov chain's stationary distribution, and an **exact expected-race-time formula** that the Monte Carlo must reproduce.

**Extensions.** §5.8 adds an exact-vs-Monte-Carlo check of the decision quantity across the whole $P_{GS}$ range; §5.9 and §6.6 add a **two-car team model** (next-event discrete-event simulation of a shared pit box); §6.7 **calibrates** the model on 16 real Singapore races (2008–2025) with held-out validation; §6.4 and §6.8 study **sensitivity and parameter uncertainty**; §6.5 measures the **speed-up** of the vectorised simulator.

## 2. Problem and question

Formula 1 teams trade the pace of fresh tyres against the time lost in the pit lane. A stop taken under SC or VSC is cheaper, because cars on track slow down while the pit lane keeps its speed limit. Singapore is the extreme case: every race from 2008 to 2023 had at least one Safety Car, but 2024 and 2025 had none. The 2025 pit-lane speed limit rise (60 → 80 km/h) also cut the pit loss by roughly 6 s.

**Question.** For a 62-lap Singapore race, how does the expected race-time difference (with 95% CI) between the best one-stop, two-stop and pit-if-neutralised strategies change as the per-lap SC onset probability $P_{GS}$ varies, and at what $P_{GS}$ does the preferred strategy switch?

**Scope.** One car, no traffic or positions (objective: expected race time); dry race; no red flags; no refuelling.

## 3. Model, assumptions and equations

### 3.1 Assumptions
| # | Assumption | Why | Unlocks |
|---|---|---|---|
| H1 | Lap-discrete time, fixed 62 laps, no red flags | Status, stops and tyre life are logged per lap | finite-horizon DTMC |
| H2 | Track status is exogenous (independent of our strategy) | one car barely affects incident risk | valid CRN pairing |
| H3 | Markov and time-homogeneous after lap 1; separate lap-1 onset | lap-1 incidents are over-represented | exact marginals via $P^\ell$ |
| H4 | SC lasts at least $m$ laps (phase-type), then geometric | real SCs last ≈2–8 laps | closed-form duration law |
| H5 | VSC/SC end in Green | escalations are rare | renewal-reward stationary law |
| H6 | Under VSC/SC lap time $=k_s t_0$, set by the neutralisation | cars follow the SC / VSC delta | SC laps cancel in paired differences |
| H7 | Green lap time is additive and separable | standard in race simulation | linearity of expectation |
| H8 | Smooth, self-accelerating tyre wear ODE | worn tyres overheat; no cliff inside a stint | closed form, convex stint cost, meaningful RK4 order |
| H9 | Pit lane at constant limit speed; bypassed section slows by $k_s$ | pit limiter | derived $\alpha_s$ |
| H10 | i.i.d. Gaussian lap noise, i.i.d. shifted-lognormal crew time | standard input models | CLT, exact sampler tests |
| H11 | Risk-neutral objective (expected race time) | single car, no positions | well-posed comparison |
| H12 | Race time is non-decreasing in every common random input | true by construction | CRN cannot increase variance |
| H13 | Stage 2 uses fresh seeds, shortlist fixed beforehand | avoids winner's-curse bias | unbiased final estimates |

### 3.2 Tyre wear and fuel (numerical ODE solvers)
Fuel mass on race lap $\ell$: $m(\ell)=m_0-b(\ell-1)$ [kg]. Tyre degradation $w$ [s] with tyre age $a$ [laps] in a stint starting on lap $s$:
$$\frac{dw}{da}=k_c\,(1+\beta\,m(s+a))\,(1+\eta\,w)\,\varphi(\text{status}),\qquad w(0)=0,$$
with $\varphi=1,\ \varphi_V,\ \varphi_S$ for Green, VSC, SC. For constant $\varphi$ the equation separates, giving the exact solution used for verification:
$$w(a)=\frac{e^{\eta k_c\varphi A(a)}-1}{\eta},\qquad A(a)=\big(1+\beta(m_0-b(s-1))\big)a-\tfrac{\beta b}{2}a^2 .$$
The simulator integrates one lap per RK4 step (status, and hence $\varphi$, is constant within a lap):
$$y_{n+1}=y_n+\tfrac{h}{6}(k_1+2k_2+2k_3+k_4),\qquad \text{global error }O(h^4).$$
Since the growth rate $\lambda=\eta k_c(1+\beta m)\approx 10^{-3}$ per lap, $|h\lambda|\ll 1$ and the scheme is stable at $h=1$ lap. *If $\eta=0$ the right-hand side depends only on $a$, and RK4 reduces to Simpson's rule and is exact. That is why $\eta>0$ is kept, so the order study is meaningful.*

### 3.3 Lap time
$$t_\ell=\begin{cases}t_0+\Delta_c+w_\ell+\gamma\,m(\ell)+\sigma\varepsilon_\ell, & \text{Green},\ \varepsilon_\ell\sim N(0,1)\\ k_s\,t_0, & \text{VSC/SC}\end{cases}$$
The fuel term $\gamma m(\ell)$ depends only on the lap, so it **cancels** in strategy differences. Fuel influences strategy only through the wear coupling $\beta$.

### 3.4 Track status (Markov chain)
States $\{G, V, SC_1,\dots,SC_m, SC^+\}$, where $SC_i\to SC_{i+1}$ with probability 1 and $SC_m, SC^+$ continue with $p_{SS}$. The SC duration is therefore $D=m+\text{Geom}$ with
$$E[D_S]=m+\frac{p_{SS}}{1-p_{SS}},\qquad \hat p_{SS}=\frac{\sum_i(D_i-m)}{\sum_i(D_i-m+1)}.$$
Long-run fractions by renewal-reward (H5), with onset probabilities $a=P_{GV}$ and $c=P_{GS}$:
$$\pi_G=\frac{1}{1+aE[D_V]+cE[D_S]},\quad \pi_V=aE[D_V]\pi_G,\quad \pi_S=cE[D_S]\pi_G .$$
The exact lap marginals are $p(\ell)=p(1)P^{\ell-1}$. Transition probabilities are estimated by maximum likelihood $\hat P_{ij}=n_{ij}/\sum_k n_{ik}$, with Beta credible intervals.

### 3.5 Pit loss (derived from geometry)
Green net transit loss: $\Delta T_G=T_{pit}-T_{bypass}$. Under neutralisation $s$ the bypassed section takes $k_sT_{bypass}$, but the speed-limited lane does not change:
$$\Delta T_s=(1-\alpha_s)\Delta T_G,\qquad \alpha_s=\frac{(k_s-1)\,T_{bypass}}{\Delta T_G}.$$
Total in-lap loss $=(1-\alpha_s)\Delta T_G + C$, with stationary time $C=c_0+e^{\mu+\sigma Z}$ (shifted lognormal, never scaled).

### 3.6 Analytic results used as verification references
* **One-stop optimum** (linear wear, $\eta=\beta=0$): $f(L)$ is convex, with $f''=k_1+k_2$, and
  $L^*=\dfrac{k_2N+(\Delta_2-\Delta_1)+(k_1-k_2)/2}{k_1+k_2}$.
* **One vs two stops, Green, equal stints:** a two-stop wins iff $kN^2/12>\Delta T_G+E[C]$.
* **Exact expectation for fixed strategies** (H2–H7, $\eta=\beta=0$, $\varphi=1$):
  $E[T]=\sum_\ell \big(p_G(\ell)\,g_\ell+p_V(\ell)k_Vt_0+p_S(\ell)k_St_0\big)+\sum_{L\in\mathcal L}\big(E[C]+\Delta T_G(1-\alpha_V p_V(L)-\alpha_S p_S(L))\big)$.

### 3.7 Monte Carlo with common random numbers
$\operatorname{Var}(T_B-T_A)=\operatorname{Var}T_A+\operatorname{Var}T_B-2\operatorname{Cov}(T_A,T_B)$. Race times are non-decreasing in the shared inputs (H12), so $\operatorname{Cov}\ge 0$ and CRN cannot increase the variance. Each source has its own random stream, so strategies with different numbers of stops still use the same scenario. CI: $\bar D\pm z\,s_D/\sqrt n$, with a Bonferroni-adjusted $z$ for multiple comparisons. The required sample size is $n\ge(z s_D/h^*)^2$.

### 3.8 Concepts applied
| Equation / algorithm | Concept |
|---|---|
| Tyre-wear ODE, RK4 (and Euler/RK2 for comparison), error order $O(h^4)$ | Numerical ODE solvers |

## 4. Implementation

The model code is below, one cell per component (source of truth: `src/f1sim.py` in the repository). Units: seconds [s], kilograms [kg], laps. Status codes: 0 Green, 1 VSC, 2 SC.

### 4.1 Header and imports

```python
"""
f1sim - single-car Formula 1 pit-stop strategy simulator.

Model summary (see the notebook for derivations):
  * Tyre wear ODE per lap, solved with RK4:
        dw/dl = k_c * (1 + beta * m(l)) * (1 + eta * w) * phi(status)        [s/lap]
  * Fuel mass m(l) = m0 - b * (l - 1)                                         [kg]
  * Green lap time  t = t0 + offset_c + w + gamma * m + eps,  eps ~ N(0, sigma^2)
  * VSC / SC lap time t = k_s * t0  (set by the neutralisation, not the car)
  * Track status: lap-wise DTMC over {Green, VSC, SC_1..SC_m, SC+} (phase-type
    minimum Safety-Car length), separate lap-1 onset probability.
  * Pit loss on the in-lap: (1 - alpha_s) * dT_G + crew time, with
    alpha_s = (k_s - 1) * T_bypass / dT_G derived from pit-lane geometry and
    crew time ~ shifted lognormal.
  * Common random numbers: independent streams for status, lap noise and crew
    time (numpy SeedSequence), so every strategy sees the same race scenario.

Units: time in seconds [s], fuel in kilograms [kg], laps are 1-based integers.
Status codes: 0 = Green, 1 = VSC, 2 = SC.
"""

import copy
import math
import time

import numpy as np

GREEN, VSC, SC = 0, 1, 2
STATUS_NAMES = ("Green", "VSC", "SC")
DRY_COMPOUNDS = ("SOFT", "MEDIUM", "HARD")
```

### 4.2 Parameters

```python
def default_params():
    """Baseline parameter set for a 62-lap Singapore race.

    Every value is tagged in PARAM_SOURCES as 'literature', 'public data' or
    'assumed (calibrate with FastF1)'. Returns a new dict each call.
    """
    p = dict(
        n_laps=62,                      # [laps] 2025 Singapore race distance
        t0_s=96.0,                      # [s] base clean lap time, new tyre, no fuel
        offset_s={"SOFT": 0.0, "MEDIUM": 0.35, "HARD": 0.70},   # [s] compound pace offset
        k_s_per_lap={"SOFT": 0.100, "MEDIUM": 0.060, "HARD": 0.035},  # [s/lap] wear rate
        eta_per_s=0.02,                 # [1/s] wear self-acceleration (worn tyres wear faster)
        beta_per_kg=0.004,              # [1/kg] extra wear per kg of fuel
        m0_kg=100.0,                    # [kg] start fuel
        burn_kg_per_lap=1.6,            # [kg/lap] fuel burn
        gamma_s_per_kg=0.03,            # [s/kg] lap-time penalty per kg of fuel
        sigma_s=0.35,                   # [s] lap-time noise standard deviation
        # neutralisations
        k_vsc=1.40,                     # [-] VSC lap time / reference lap time
        k_sc=1.60,                      # [-] SC lap time / reference lap time
        phi_vsc=0.50,                   # [-] tyre-wear factor under VSC
        phi_sc=0.25,                    # [-] tyre-wear factor under SC
        # Markov chain (per-lap probabilities)
        p_gv=0.012,                     # [-] Green -> VSC onset
        p_gs=0.020,                     # [-] Green -> SC onset
        p_lap1_sc=0.15,                 # [-] SC onset on lap 1 (start incidents)
        p_vv=0.60,                      # [-] VSC persistence
        sc_min_laps=3,                  # [laps] minimum Safety-Car length m
        p_ss=0.50,                      # [-] SC persistence after the minimum
        # pit lane geometry (2025, 80 km/h limit)
        t_pit_s=36.0,                   # [s] pit entry -> exit through the lane (incl. no stop)
        t_bypass_s=15.0,                # [s] same section driven on track at racing speed
        # stationary crew time T = shift + exp(mu + sigma * Z)
        crew_shift_s=2.0,               # [s] physical minimum stop
        crew_mu=-0.51,                  # [-] log-scale location (median extra ~0.6 s)
        crew_sigma=0.50,                # [-] log-scale spread
        random=True,                    # False -> deterministic benchmark mode
    )
    return derive_alphas(p)


PARAM_SOURCES = {
    "n_laps": "public data (2025 race report)",
    "k_vsc, k_sc": "literature (Heilmeier et al. 2020, ~140% / ~160%)",
    "phi_vsc, phi_sc": "literature (Heilmeier et al. 2020, tyre ageing -50% / -75%)",
    "p_gs, p_gv, p_lap1_sc": "public data, rough (~7 SC onsets in ~370 laps, 2018-2025); calibrate with FastF1",
    "t_pit_s, t_bypass_s": "assumed (calibrate with FastF1 2025 PitInTime/PitOutTime)",
    "crew_*": "assumed, informed by published stationary times (DHL)",
    "t0_s, offset_s, k_s_per_lap, eta, beta, gamma, sigma": "assumed (calibrate with FastF1 clean stints)",
}


def derive_alphas(p):
    """Derive VSC/SC net-transit-loss reductions from pit-lane geometry.

    dT_G = T_pit - T_bypass; under neutralisation the bypassed section takes
    k_s * T_bypass while the speed-limited lane is unchanged, so
    alpha_s = (k_s - 1) * T_bypass / dT_G.  Modifies and returns p.
    """
    d_green = p["t_pit_s"] - p["t_bypass_s"]          # [s]
    assert d_green > 0, "pit lane must be slower than the track"
    p["dT_green_s"] = d_green
    p["alpha_vsc"] = (p["k_vsc"] - 1.0) * p["t_bypass_s"] / d_green
    p["alpha_sc"] = (p["k_sc"] - 1.0) * p["t_bypass_s"] / d_green
    return p


def with_params(p, **changes):
    """Return a deep copy of p with changes applied (alphas re-derived unless overridden)."""
    q = copy.deepcopy(p)
    q.update(changes)
    if "alpha_sc" not in changes and "alpha_vsc" not in changes:
        derive_alphas(q)
    return q
```

### 4.3 Numerical ODE solvers

```python
def euler_solve(f, y0, t0, t1, n_steps):
    """Forward Euler: returns y(t1) after n_steps equal steps (global error O(h))."""
    h = (t1 - t0) / n_steps
    y = np.asarray(y0, dtype=float) if np.ndim(y0) else float(y0)
    t = t0
    for _ in range(n_steps):
        y = y + h * f(t, y)
        t += h
    return y


def rk2_solve(f, y0, t0, t1, n_steps):
    """Midpoint RK2: returns y(t1) (global error O(h^2))."""
    h = (t1 - t0) / n_steps
    y = np.asarray(y0, dtype=float) if np.ndim(y0) else float(y0)
    t = t0
    for _ in range(n_steps):
        k1 = f(t, y)
        k2 = f(t + h / 2, y + h / 2 * k1)
        y = y + h * k2
        t += h
    return y


def rk4_step(f, t, y, h):
    """One classical RK4 step of size h (local error O(h^5))."""
    k1 = f(t, y)
    k2 = f(t + h / 2, y + h / 2 * k1)
    k3 = f(t + h / 2, y + h / 2 * k2)
    k4 = f(t + h, y + h * k3)
    return y + h / 6.0 * (k1 + 2 * k2 + 2 * k3 + k4)


def rk4_solve(f, y0, t0, t1, n_steps):
    """Classical RK4: returns y(t1) (global error O(h^4)). y0 may be a float or array."""
    h = (t1 - t0) / n_steps
    y = np.asarray(y0, dtype=float) if np.ndim(y0) else float(y0)
    t = t0
    for _ in range(n_steps):
        y = rk4_step(f, t, y, h)
        t += h
    return y


def wear_rhs(p, compound, race_lap0, phi):
    """Right-hand side dw/da of the tyre-wear ODE for one lap.

    a is lap time-like (age measured in laps), race_lap0 is the race lap at
    a = 0 of this step, phi is the neutralisation factor. Works on arrays.
    Returns f(a, w) in [s/lap].
    """
    k = p["k_s_per_lap"][compound]
    beta, eta = p["beta_per_kg"], p["eta_per_s"]
    m0, b = p["m0_kg"], p["burn_kg_per_lap"]

    def f(a, w):
        fuel_kg = m0 - b * (race_lap0 - 1 + a)
        return k * (1.0 + beta * fuel_kg) * (1.0 + eta * w) * phi
    return f


def wear_closed_form(p, compound, stint_start_lap, age_laps, phi=1.0):
    """Exact solution of the wear ODE for constant phi (verification reference).

    w(a) = (exp(eta * k * phi * A(a)) - 1) / eta,
    A(a) = (1 + beta*(m0 - b*(s-1))) * a - beta*b*a^2/2.   Returns w [s].
    """
    k = p["k_s_per_lap"][compound]
    beta, eta = p["beta_per_kg"], p["eta_per_s"]
    m0, b = p["m0_kg"], p["burn_kg_per_lap"]
    a = np.asarray(age_laps, dtype=float)
    A = (1 + beta * (m0 - b * (stint_start_lap - 1))) * a - beta * b * a ** 2 / 2
    if eta == 0:
        return k * phi * A
    return (np.exp(eta * k * phi * A) - 1.0) / eta
```

### 4.4 Random variate generation

```python
def box_muller(n, rng):
    """n standard normals from uniforms by the Box-Muller transform."""
    m = (n + 1) // 2
    u1 = 1.0 - rng.random(m)            # in (0, 1], avoids log(0)
    u2 = rng.random(m)
    r = np.sqrt(-2.0 * np.log(u1))
    z = np.concatenate([r * np.cos(2 * np.pi * u2), r * np.sin(2 * np.pi * u2)])
    return z[:n]


def box_muller_from_uniforms(u1, u2):
    """Box-Muller on given uniform arrays (u1 in (0, 1]). Returns (z1, z2)."""
    r = np.sqrt(-2.0 * np.log(u1))
    return r * np.cos(2 * np.pi * u2), r * np.sin(2 * np.pi * u2)


def crew_time_params(p):
    """(shift [s], mu, sigma) of the shifted-lognormal stationary crew time."""
    return p["crew_shift_s"], p["crew_mu"], p["crew_sigma"]


def crew_time_from_normal(z, p):
    """Map standard normals z to stationary crew times [s] (exact inverse transform)."""
    return p["crew_shift_s"] + np.exp(p["crew_mu"] + p["crew_sigma"] * z)


def sample_crew_time(n, p, rng):
    """n stationary crew times [s] from the shifted lognormal."""
    return crew_time_from_normal(box_muller(n, rng), p)
```

### 4.5 Track-status Markov chain

```python
def lumped_matrix(p):
    """3x3 matrix over (Green, VSC, SC) with the same mean sojourn times.

    Used for reporting and for the stationarity check; the simulator uses the
    expanded phase-type chain from expanded_matrix().
    """
    a, c = p["p_gv"], p["p_gs"]
    mean_sc = sc_mean_duration(p)                       # [laps]
    r_sc = 1.0 / mean_sc
    return np.array([[1 - a - c, a, c],
                     [1 - p["p_vv"], p["p_vv"], 0.0],
                     [r_sc, 0.0, 1 - r_sc]])


def expanded_matrix(p):
    """Phase-type chain: states [G, V, SC_1..SC_m, SC+].

    SC_i -> SC_{i+1} with probability 1 enforces a minimum length m; SC_m and
    SC+ continue with p_ss or return to Green. Returns (P, collapse) where
    collapse[i] maps an expanded state to 0/1/2.
    """
    m = int(p["sc_min_laps"])
    n = 2 + m + 1
    P = np.zeros((n, n))
    a, c = p["p_gv"], p["p_gs"]
    P[0, 0], P[0, 1], P[0, 2] = 1 - a - c, a, c
    P[1, 0], P[1, 1] = 1 - p["p_vv"], p["p_vv"]
    for i in range(m - 1):
        P[2 + i, 3 + i] = 1.0
    last = 2 + m - 1
    for s in (last, n - 1):
        P[s, n - 1] = p["p_ss"]
        P[s, 0] = 1 - p["p_ss"]
    collapse = np.array([0, 1] + [2] * (m + 1))
    assert P.min() >= 0 and np.allclose(P.sum(1), 1.0)
    return P, collapse


def lap1_row(p):
    """Distribution of the expanded state on lap 1 (start from the grid in Green)."""
    P, _ = expanded_matrix(p)
    row = np.zeros(P.shape[0])
    q = p["p_lap1_sc"]
    row[0] = 1 - q - p["p_gv"]
    row[1] = p["p_gv"]
    row[2] = q
    return row


def sc_mean_duration(p):
    """E[D_SC] = m + p_ss / (1 - p_ss) laps (phase-type minimum + geometric tail)."""
    return p["sc_min_laps"] + p["p_ss"] / (1 - p["p_ss"])


def stationary_distribution(P):
    """Solve pi = pi P, sum(pi) = 1 by least squares on the augmented system."""
    n = P.shape[0]
    A = np.vstack([P.T - np.eye(n), np.ones(n)])
    b = np.zeros(n + 1)
    b[-1] = 1.0
    return np.linalg.lstsq(A, b, rcond=None)[0]


def renewal_fractions(p):
    """Closed-form long-run fractions (Green, VSC, SC) when neutralisations end in Green.

    pi_G = 1 / (1 + a E[D_V] + c E[D_S]), pi_V = a E[D_V] pi_G, pi_S = c E[D_S] pi_G.
    """
    ev = 1.0 / (1 - p["p_vv"])
    es = sc_mean_duration(p)
    g = 1.0 / (1 + p["p_gv"] * ev + p["p_gs"] * es)
    return np.array([g, p["p_gv"] * ev * g, p["p_gs"] * es * g])


def sample_next_state(cum_rows, state, u):
    """Inverse-transform step for a DTMC, vectorised over replications."""
    c = cum_rows[state]                                   # (n, k) cumulative rows
    nxt = (u[:, None] > c).sum(axis=1)
    return np.minimum(nxt, c.shape[1] - 1)


def simulate_track_status(P, n_steps, rng, start=0):
    """Simulate a DTMC path of n_steps from matrix P by inverse transform."""
    P = np.asarray(P, dtype=float)
    cum = np.cumsum(P, axis=1)
    u = rng.random(n_steps)
    out = np.empty(n_steps, dtype=int)
    s = start
    for i in range(n_steps):
        out[i] = s
        s = min(int(np.searchsorted(cum[s], u[i], side="right")), P.shape[0] - 1)
    return out


def status_probabilities(p):
    """Exact marginal P(status on lap l) for l = 1..N, shape (N, 3)."""
    P, collapse = expanded_matrix(p)
    row = lap1_row(p)
    out = np.zeros((p["n_laps"], 3))
    for l in range(p["n_laps"]):
        for s in range(3):
            out[l, s] = row[collapse == s].sum()
        row = row @ P
    return out


def mle_transition_matrix(counts):
    """Maximum-likelihood transition matrix: row-normalised transition counts."""
    C = np.asarray(counts, dtype=float)
    return C / np.maximum(C.sum(axis=1, keepdims=True), 1.0)


def beta_interval(successes, trials, level=0.95, prior=(1.0, 1.0), n_draws=200000, seed=0):
    """Posterior mean and equal-tailed credible interval for a Bernoulli probability."""
    rng = np.random.default_rng(seed)
    a, b = prior[0] + successes, prior[1] + trials - successes
    draws = rng.beta(a, b, n_draws)
    lo, hi = np.quantile(draws, [(1 - level) / 2, (1 + level) / 2])
    return a / (a + b), lo, hi


def mle_sc_persistence(durations, m=None):
    """Phase-type SC fit: m = min duration, p_ss = sum(D-m) / sum(D-m+1)."""
    d = np.asarray(durations, dtype=float)
    m = int(d.min()) if m is None else m
    ex = d - m
    return m, ex.sum() / (ex + 1).sum()
```

### 4.6 Pit loss

```python
def pit_loss(state, crew_time_s, p):
    """Total time lost on the in-lap for one stop [s].

    (1 - alpha_state) * dT_G + crew time; the stationary time is never scaled.
    Works with scalar or array state / crew time.
    """
    alpha = np.choose(np.asarray(state), [0.0, p["alpha_vsc"], p["alpha_sc"]])
    return (1.0 - alpha) * p["dT_green_s"] + crew_time_s
```

### 4.7 Strategies

```python
def strategy(pit_laps, compounds, reactive_window=None, name=None):
    """Build a strategy dict.

    pit_laps: planned in-laps, compounds: one per stint. reactive_window=(a, b)
    makes the FIRST stop move earlier to the first lap in [a, b] that is under
    VSC/SC (pit if neutralised).
    """
    s = dict(pit_laps=list(pit_laps), compounds=list(compounds), reactive_window=reactive_window)
    if name is None:
        name = "%d-stop %s @%s%s" % (len(pit_laps), "-".join(c[0] for c in compounds),
                                     ",".join(str(l) for l in pit_laps),
                                     "" if reactive_window is None else " react[%d-%d]" % tuple(reactive_window))
    s["name"] = name
    return s


def validate_strategy(s, p):
    """True if the strategy is legal: >= 2 dry compounds, increasing in-laps inside the race."""
    laps, comps = s["pit_laps"], s["compounds"]
    n = p["n_laps"]
    if len(comps) != len(laps) + 1 or len(laps) == 0:
        return False
    if any(c not in DRY_COMPOUNDS for c in comps) or len(set(comps)) < 2:
        return False
    if any(not (1 <= l <= n - 1) for l in laps) or any(b <= a for a, b in zip(laps, laps[1:])):
        return False
    w = s.get("reactive_window")
    if w is not None:
        if not (1 <= w[0] <= w[1] <= laps[0]):
            return False
    return True
```

### 4.8 Scenarios (common random numbers) and the race simulator

```python
MAX_STOPS = 3


def draw_scenarios(n_reps, p, seed):
    """Draw n_reps race scenarios from three independent streams.

    Returns dict with 'status' (n, N) int, 'z_lap' (n, N) lap-noise normals and
    'z_crew' (n, MAX_STOPS) normals for crew times. The same scenarios are
    re-used for every strategy (common random numbers).
    """
    N = p["n_laps"]
    ss_status, ss_noise, ss_crew = np.random.SeedSequence(seed).spawn(3)
    g_status = np.random.default_rng(ss_status)
    g_noise = np.random.default_rng(ss_noise)
    g_crew = np.random.default_rng(ss_crew)

    if p.get("random", True):
        P, collapse = expanded_matrix(p)
        cum = np.cumsum(P, axis=1)
        u = g_status.random((n_reps, N))
        cum1 = np.cumsum(lap1_row(p))
        state = np.minimum((u[:, 0][:, None] > cum1[None, :]).sum(1), P.shape[0] - 1)
        status = np.empty((n_reps, N), dtype=np.int8)
        status[:, 0] = collapse[state]
        for l in range(1, N):
            state = sample_next_state(cum, state, u[:, l])
            status[:, l] = collapse[state]
        z_lap = box_muller(n_reps * N, g_noise).reshape(n_reps, N)
        z_crew = box_muller(n_reps * MAX_STOPS, g_crew).reshape(n_reps, MAX_STOPS)
    else:
        status = np.zeros((n_reps, N), dtype=np.int8)
        z_lap = np.zeros((n_reps, N))
        z_crew = np.zeros((n_reps, MAX_STOPS))
    return dict(status=status, z_lap=z_lap, z_crew=z_crew)


def _first_stop_laps(s, status):
    """Actual first in-lap per replication, applying the reactive rule if any."""
    n = status.shape[0]
    first = np.full(n, s["pit_laps"][0], dtype=int)
    w = s.get("reactive_window")
    if w is not None:
        a, b = w
        seg = status[:, a - 1:b] != GREEN                 # laps a..b
        hit = seg.any(axis=1)
        first[hit] = a + seg[hit].argmax(axis=1)
    return first


def simulate_scenarios(s, p, scen, return_details=False):
    """Vectorised race simulation of strategy s over all scenarios.

    Returns total race times [s] (n,), or a dict of per-lap arrays when
    return_details=True.
    """
    status = scen["status"]
    n, N = status.shape
    noise_sd = p["sigma_s"] if p.get("random", True) else 0.0
    crew = crew_time_from_normal(scen["z_crew"], p) if p.get("random", True) else \
        np.full((n, MAX_STOPS), p["crew_shift_s"] + math.exp(p["crew_mu"]))

    first = _first_stop_laps(s, status)
    later = np.array(s["pit_laps"][1:], dtype=int)
    stint = np.zeros(n, dtype=int)
    w = np.zeros(n)                                       # [s] tyre degradation
    age = np.zeros(n)                                     # [laps]
    comps = s["compounds"]
    lap_t = np.zeros((n, N))
    if return_details:
        fuel_hist, age_hist, w_hist = np.zeros((n, N)), np.zeros((n, N)), np.zeros((n, N))
    phi_of = np.array([1.0, p["phi_vsc"], p["phi_sc"]])
    k_of = np.array([1.0, p["k_vsc"], p["k_sc"]])
    pit_laps_taken = [[] for _ in range(n)] if return_details else None

    for l in range(1, N + 1):
        st = status[:, l - 1]
        fuel = p["m0_kg"] - p["burn_kg_per_lap"] * (l - 1)          # [kg]
        offset = np.array([p["offset_s"][c] for c in comps])[stint]
        green_t = p["t0_s"] + offset + w + p["gamma_s_per_kg"] * fuel + noise_sd * scen["z_lap"][:, l - 1]
        t = np.where(st == GREEN, green_t, k_of[st] * p["t0_s"])
        if return_details:
            fuel_hist[:, l - 1], age_hist[:, l - 1], w_hist[:, l - 1] = fuel, age, w
        # integrate wear over this lap (one RK4 step of 1 lap, per compound)
        phi = phi_of[st]
        new_w = np.empty(n)
        for ci, c in enumerate(comps):
            idx = stint == ci
            if idx.any():
                f = wear_rhs(p, c, l, phi[idx])
                new_w[idx] = rk4_step(f, 0.0, w[idx], 1.0)
        w, age = new_w, age + 1
        # pit stop on this in-lap?
        stop = np.where(stint == 0, first == l, False)
        if len(later):
            for j, L in enumerate(later):
                stop |= (stint == j + 1) & (l == L)
        if stop.any():
            idx = np.nonzero(stop)[0]
            t[idx] += pit_loss(st[idx], crew[idx, stint[idx]], p)
            stint[idx] += 1
            w[idx], age[idx] = 0.0, 0.0
            if return_details:
                for i in idx:
                    pit_laps_taken[i].append(l)
        lap_t[:, l - 1] = t
    total = lap_t.sum(axis=1)
    if not return_details:
        return total
    return dict(total_time=total, lap_times=lap_t, status=status, fuel_kg=fuel_hist,
                tyre_age=age_hist, wear_s=w_hist, pit_laps=pit_laps_taken)


def simulate_race(s, p, seed):
    """One race for one strategy (scenario derived from seed). Returns a dict of arrays."""
    d = simulate_scenarios(s, p, draw_scenarios(1, p, seed), return_details=True)
    return dict(total_time=float(d["total_time"][0]), lap_times=d["lap_times"][0],
                status=d["status"][0].astype(int), fuel_kg=d["fuel_kg"][0],
                tyre_age=d["tyre_age"][0], pit_laps=d["pit_laps"][0])


def run_paired(strategies, p, n_reps, seed):
    """Total race times (n_reps, n_strategies) with every scenario re-used (CRN)."""
    scen = draw_scenarios(n_reps, p, seed)
    return np.column_stack([simulate_scenarios(s, p, scen) for s in strategies])


def simulate_race_loop(s, p, scen, i):
    """Plain-Python single-replication simulator (baseline for the speed-up study).

    Uses scenario row i, so it must reproduce simulate_scenarios exactly.
    """
    N = p["n_laps"]
    status = scen["status"][i]
    first = int(_first_stop_laps(s, scen["status"][i:i + 1])[0])
    stops = [first] + list(s["pit_laps"][1:])
    stint, w, total = 0, 0.0, 0.0
    phi_of = (1.0, p["phi_vsc"], p["phi_sc"])
    k_of = (1.0, p["k_vsc"], p["k_sc"])
    rand = p.get("random", True)
    for l in range(1, N + 1):
        st = int(status[l - 1])
        c = s["compounds"][stint]
        fuel = p["m0_kg"] - p["burn_kg_per_lap"] * (l - 1)
        if st == GREEN:
            t = p["t0_s"] + p["offset_s"][c] + w + p["gamma_s_per_kg"] * fuel
            if rand:
                t += p["sigma_s"] * scen["z_lap"][i, l - 1]
        else:
            t = k_of[st] * p["t0_s"]
        w = float(rk4_step(wear_rhs(p, c, l, phi_of[st]), 0.0, w, 1.0))
        if stint < len(stops) and l == stops[stint]:
            z = scen["z_crew"][i, stint]
            crew = p["crew_shift_s"] + math.exp(p["crew_mu"] + p["crew_sigma"] * z) if rand \
                else p["crew_shift_s"] + math.exp(p["crew_mu"])
            t += float(pit_loss(st, crew, p))
            stint += 1
            w = 0.0
        total += t
    return total
```

### 4.9 Two-car team model (next-event discrete-event simulation)

```python
def team_params(p, **changes):
    """Team-model settings added to a copy of p (all times in seconds).

    grid_gap_s   car B starts this far behind car A
    follow_gap_s minimum gap when a car is stuck behind its team-mate on track
    sc_gap_s     gap the field bunches up to behind the Safety Car
    box_turn_s   crew turnaround before the box can serve the second car
    rho_noise    correlation of the two cars' lap noise (shared track evolution)
    interact     False -> the cars ignore each other (independent single-car races)
    """
    q = copy.deepcopy(p)
    q.setdefault("grid_gap_s", 0.8)
    q.setdefault("follow_gap_s", 0.6)
    q.setdefault("sc_gap_s", 0.8)
    q.setdefault("box_turn_s", 1.5)
    q.setdefault("rho_noise", 0.7)
    q.setdefault("interact", True)
    q.update(changes)
    return q


def draw_team_scenarios(n_reps, p, seed):
    """Scenarios for two cars. Car A uses exactly draw_scenarios(seed); car B gets
    its own lap-noise and crew streams (lap noise correlated with A by rho_noise)
    and shares A's track status."""
    q = team_params(p)
    a = draw_scenarios(n_reps, q, seed)
    ss = np.random.SeedSequence([seed, 2])
    g_noise, g_crew = [np.random.default_rng(s) for s in ss.spawn(2)]
    if q.get("random", True):
        rho = q["rho_noise"]
        z_ind = box_muller(a["z_lap"].size, g_noise).reshape(a["z_lap"].shape)
        z_b = rho * a["z_lap"] + math.sqrt(1 - rho ** 2) * z_ind
        zc_b = box_muller(a["z_crew"].size, g_crew).reshape(a["z_crew"].shape)
    else:
        z_b, zc_b = np.zeros_like(a["z_lap"]), np.zeros_like(a["z_crew"])
    b = dict(status=a["status"], z_lap=z_b, z_crew=zc_b)
    return a, b


class _Car:
    """State of one car inside the event loop."""

    def __init__(self, name, s, p, scen, i):
        self.name, self.s, self.p = name, s, p
        self.status = scen["status"][i]
        self.z_lap, self.z_crew = scen["z_lap"][i], scen["z_crew"][i]
        first = int(_first_stop_laps(s, scen["status"][i:i + 1])[0])
        self.stops = [first] + list(s["pit_laps"][1:])
        self.lap, self.stint, self.w = 0, 0, 0.0          # completed laps, stint index, wear [s]
        self.lap_start = 0.0                                # race clock at the start of the current lap
        self.line_times = []                                # race clock at each line crossing
        self.stack_wait = 0.0                               # [s] waited behind the team-mate in the box
        self.traffic = 0.0                                  # [s] lost stuck behind the team-mate
        self.pending_line = None                            # scheduled crossing time of the current lap
        self.in_pit = False

    def natural_lap(self, l):
        """Lap time [s] of lap l ignoring the other car (identical to the single-car model)."""
        p, st = self.p, int(self.status[l - 1])
        c = self.s["compounds"][self.stint]
        if st == GREEN:
            fuel = p["m0_kg"] - p["burn_kg_per_lap"] * (l - 1)
            t = p["t0_s"] + p["offset_s"][c] + self.w + p["gamma_s_per_kg"] * fuel
            if p.get("random", True):
                t += p["sigma_s"] * self.z_lap[l - 1]
        else:
            t = (p["k_vsc"] if st == VSC else p["k_sc"]) * p["t0_s"]
        phi = (1.0, p["phi_vsc"], p["phi_sc"])[st]
        self.w = float(rk4_step(wear_rhs(p, c, l, phi), 0.0, self.w, 1.0))
        return t

    def crew_time(self):
        p = self.p
        z = self.z_crew[self.stint] if p.get("random", True) else 0.0
        return p["crew_shift_s"] + math.exp(p["crew_mu"] + p["crew_sigma"] * z) if p.get("random", True) \
            else p["crew_shift_s"] + math.exp(p["crew_mu"])


def simulate_team_race(sA, sB, p, scen_a, scen_b, i=0, log=None):
    """Two team-mates, one pit box, next-event time advance.

    Events (time-ordered with a heap): LINE (car crosses the line, starts the
    next lap), BOX_ARRIVE, BOX_DONE. Interactions:
      * double-stacking: one box, first come first served; the second car waits
        until the first car's service ends plus a crew turnaround box_turn_s;
      * no passing on track: a car that starts a lap behind its team-mate cannot
        cross the line less than follow_gap_s behind it (traffic on rejoining is
        the same rule applied to the out-lap);
      * Safety-Car bunching: on SC laps the trailing car closes up to sc_gap_s.
    With interact=False none of these apply and each car reproduces
    simulate_scenarios exactly. Returns a dict of totals and delays [s].
    """
    import heapq
    q = team_params(p)
    cars = [_Car("A", sA, q, scen_a, i), _Car("B", sB, q, scen_b, i)]
    N = q["n_laps"]
    box_free, busy, queue, ev, seq = 0.0, False, [], [], 0

    def push(t, kind, k):
        nonlocal seq
        heapq.heappush(ev, (t, seq, kind, k))
        seq += 1

    def start_lap(k, t):
        """Car k crosses the line at t and begins lap car.lap + 1."""
        car = cars[k]
        car.lap_start = t
        l = car.lap + 1
        d = car.natural_lap(l)
        if car.stint < len(car.stops) and l == car.stops[car.stint]:
            st = int(car.status[l - 1])
            alpha = (0.0, q["alpha_vsc"], q["alpha_sc"])[st]
            car.transit = (1 - alpha) * q["dT_green_s"]
            car.in_pit = True
            car.pending_line = None
            push(t + d + car.transit / 2, "BOX_ARRIVE", k)
        else:
            car.in_pit = False
            car.pending_line = t + d
            push(t + d, "LINE", k)

    def serve(k, t):
        nonlocal box_free, busy
        busy = True
        c = cars[k].crew_time()
        box_free = t + c + q["box_turn_s"]
        push(t + c, "BOX_DONE", k)

    for k, t0 in ((0, 0.0), (1, q["grid_gap_s"])):
        start_lap(k, t0)
    while ev:
        t, _, kind, k = heapq.heappop(ev)
        car, other = cars[k], cars[1 - k]
        if kind == "BOX_ARRIVE":
            if q["interact"] and (busy or queue):
                queue.append((k, t))                        # team-mate still in the box: wait
            elif q["interact"] and t < box_free:
                car.stack_wait += box_free - t              # crew still resetting after the team-mate
                serve(k, box_free)
            else:
                serve(k, t)
        elif kind == "BOX_DONE":
            busy = False
            car.stint += 1
            car.w = 0.0
            car.in_pit = False
            car.pending_line = t + car.transit / 2
            push(car.pending_line, "LINE", k)
            if queue:
                j, t_arr = queue.pop(0)
                t_start = max(t + q["box_turn_s"], t_arr)
                cars[j].stack_wait += t_start - t_arr
                serve(j, t_start)
        elif kind == "LINE":
            if t != car.pending_line:
                continue                                    # superseded by a re-scheduled crossing
            l = car.lap + 1
            if q["interact"] and other.lap == car.lap and other.pending_line is not None \
                    and other.lap_start < car.lap_start and not other.in_pit:
                st = int(car.status[l - 1])
                gap = q["sc_gap_s"] if st == SC else q["follow_gap_s"]
                limit = other.pending_line + gap
                if st == SC and t > limit:
                    t_new = limit                           # bunching behind the Safety Car
                else:
                    t_new = max(t, limit)                   # cannot pass the team-mate on track
                if t_new != t:
                    car.traffic += t_new - t
                    car.pending_line = t_new
                    push(t_new, "LINE", k)
                    continue
            car.lap += 1
            car.line_times.append(t)
            if log is not None:
                log.append((t, car.name, car.lap))
            if car.lap < N:
                start_lap(k, t)
            else:
                car.pending_line = None
    tA = cars[0].line_times[-1]
    tB = cars[1].line_times[-1] - q["grid_gap_s"]
    return dict(total_a=tA, total_b=tB, stack_delay_a=cars[0].stack_wait, stack_delay_b=cars[1].stack_wait,
                traffic_a=cars[0].traffic, traffic_b=cars[1].traffic)


def simulate_two_car_race(sA, sB, p, seed):
    """One two-car race from a seed (car A sees the same scenario as simulate_race(sA, p, seed))."""
    a, b = draw_team_scenarios(1, p, seed)
    return simulate_team_race(sA, sB, p, a, b, 0)


def run_team(sA, sB, p, n_reps, seed):
    """n_reps team races on common scenarios. Returns a dict of arrays."""
    a, b = draw_team_scenarios(n_reps, p, seed)
    rows = [simulate_team_race(sA, sB, p, a, b, i) for i in range(n_reps)]
    return {k: np.array([r[k] for r in rows]) for k in rows[0]}
```

### 4.10 Analytic results (verification references)

```python
def linear_params(p):
    """Copy of p reduced to the analytically tractable case: eta = beta = 0, phi = 1."""
    return with_params(p, eta_per_s=0.0, beta_per_kg=0.0, phi_vsc=1.0, phi_sc=1.0)


def green_total_linear(s, p):
    """Exact deterministic Green race time for linear wear (eta = beta = 0), no pit loss.

    Stint of length S on compound c costs S*offset_c + k_c * S(S-1)/2 plus the
    strategy-independent base and fuel terms.
    """
    N = p["n_laps"]
    edges = [0] + list(s["pit_laps"]) + [N]
    base = N * p["t0_s"] + p["gamma_s_per_kg"] * sum(p["m0_kg"] - p["burn_kg_per_lap"] * (l - 1) for l in range(1, N + 1))
    tot = base
    for c, (a, b) in zip(s["compounds"], zip(edges, edges[1:])):
        S = b - a
        tot += S * p["offset_s"][c] + p["k_s_per_lap"][c] * S * (S - 1) / 2
    return tot


def optimal_one_stop_lap(c1, c2, p):
    """L* = [k2 N + (d2 - d1) + (k1 - k2)/2] / (k1 + k2) for the linear model (continuous)."""
    k1, k2 = p["k_s_per_lap"][c1], p["k_s_per_lap"][c2]
    d1, d2 = p["offset_s"][c1], p["offset_s"][c2]
    return (k2 * p["n_laps"] + (d2 - d1) + (k1 - k2) / 2) / (k1 + k2)


def two_stop_break_even_k(p):
    """Green break-even wear rate for 1 vs 2 equal stints: k N^2 / 12 = dT_G + E[crew]."""
    mean_crew = p["crew_shift_s"] + math.exp(p["crew_mu"] + p["crew_sigma"] ** 2 / 2)
    return 12 * (p["dT_green_s"] + mean_crew) / p["n_laps"] ** 2


def exact_expected_total(s, p):
    """Exact E[total race time] for a FIXED strategy in the linear case (eta=beta=0, phi=1).

    Uses the exact lap-status marginals from the Markov chain: SC/VSC laps take
    k_s * t0, Green laps the deterministic Green time, and each stop costs the
    expected pit loss for the status of its in-lap.
    """
    assert s.get("reactive_window") is None
    N = p["n_laps"]
    probs = status_probabilities(p)                       # (N, 3)
    mean_crew = p["crew_shift_s"] + math.exp(p["crew_mu"] + p["crew_sigma"] ** 2 / 2)
    edges = [0] + list(s["pit_laps"]) + [N]
    total = 0.0
    for c, (a, b) in zip(s["compounds"], zip(edges, edges[1:])):
        for l in range(a + 1, b + 1):
            age = l - a - 1
            green = p["t0_s"] + p["offset_s"][c] + p["k_s_per_lap"][c] * age + \
                p["gamma_s_per_kg"] * (p["m0_kg"] - p["burn_kg_per_lap"] * (l - 1))
            pr = probs[l - 1]
            total += pr[0] * green + pr[1] * p["k_vsc"] * p["t0_s"] + pr[2] * p["k_sc"] * p["t0_s"]
    for L in s["pit_laps"]:
        pr = probs[L - 1]
        total += mean_crew + p["dT_green_s"] * (pr[0] + pr[1] * (1 - p["alpha_vsc"]) + pr[2] * (1 - p["alpha_sc"]))
    return total
```

```python
def mean_ci(x, level=0.95, z=None):
    """Mean and normal-approximation CI (z = 1.96 for 95% unless given)."""
    x = np.asarray(x, dtype=float)
    z = 1.959964 if z is None else z
    h = z * x.std(ddof=1) / math.sqrt(len(x))
    return x.mean(), x.mean() - h, x.mean() + h


def bonferroni_z(n_comparisons, level=0.95):
    """Two-sided normal quantile for a Bonferroni-adjusted family-wise level."""
    from statistics import NormalDist
    return NormalDist().inv_cdf(1 - (1 - level) / (2 * n_comparisons))


def candidate_strategies(p):
    """Screening grid of legal one-stop and two-stop strategies."""
    N = p["n_laps"]
    out = []
    for c in (("MEDIUM", "HARD"), ("HARD", "MEDIUM"), ("SOFT", "HARD")):
        for L in range(10, N - 9, 2):
            out.append(strategy([L], c))
    for c in (("SOFT", "MEDIUM", "HARD"), ("MEDIUM", "HARD", "SOFT"), ("MEDIUM", "MEDIUM", "HARD")):
        for L1 in range(10, 32, 3):
            for L2 in range(L1 + 12, N - 7, 3):
                out.append(strategy([L1, L2], c))
    return out


def screen(cands, p, n_reps, seed):
    """Stage 1: mean race time of every candidate on common scenarios. Returns sorted list."""
    scen = draw_scenarios(n_reps, p, seed)
    res = [(simulate_scenarios(s, p, scen).mean(), s) for s in cands]
    res.sort(key=lambda r: r[0])
    return res


def paired_table(strats, p, n_reps, seed, ref=0, z=None):
    """Stage 2: mean and CI of every strategy, and paired differences vs strategy `ref`."""
    X = run_paired(strats, p, n_reps, seed)
    rows = []
    for j, s in enumerate(strats):
        m, lo, hi = mean_ci(X[:, j], z=z)
        dm, dlo, dhi = mean_ci(X[:, j] - X[:, ref], z=z) if j != ref else (0.0, 0.0, 0.0)
        rows.append(dict(name=s["name"], mean_s=m, ci_low_s=lo, ci_high_s=hi,
                         diff_s=dm, diff_low_s=dlo, diff_high_s=dhi))
    return rows, X


def variance_reduction_ratio(sA, sB, p, n_reps, seed):
    """Var(independent difference) / Var(CRN paired difference)."""
    X = run_paired([sA, sB], p, n_reps, seed)
    a = run_paired([sA], p, n_reps, seed + 1)[:, 0]
    b = run_paired([sB], p, n_reps, seed + 2)[:, 0]
    return np.var(a - b, ddof=1) / np.var(X[:, 0] - X[:, 1], ddof=1)


def timed(fn, *args, repeat=3):
    """Median wall time [s] of fn(*args) over `repeat` runs, and its last result."""
    ts, out = [], None
    for _ in range(repeat):
        t0 = time.perf_counter()
        out = fn(*args)
        ts.append(time.perf_counter() - t0)
    return float(np.median(ts)), out
```

### 5.6 Random variate generators (Kolmogorov–Smirnov)
Box–Muller output must be N(0,1), and crew times must follow the shifted lognormal. Each KS statistic must be below the 5% critical value $1.36/\sqrt n$, and no crew time may fall below the shift.

```python
def ks_stat(x, cdf):
    # One-sample Kolmogorov-Smirnov statistic D_n
    x = np.sort(np.asarray(x)); n = len(x); F = cdf(x)
    return max(np.max(np.arange(1, n + 1) / n - F), np.max(F - np.arange(n) / n))

norm_cdf = np.vectorize(lambda v: 0.5 * (1 + math.erf(v / math.sqrt(2))))
rng = np.random.default_rng(SEED + 2)
z = box_muller(5000, rng)
c = sample_crew_time(5000, p, rng)
shift, mu, sg = crew_time_params(p)
D_z = ks_stat(z, norm_cdf)
D_c = ks_stat(c, lambda v: norm_cdf((np.log(v - shift) - mu) / sg))
crit = 1.36 / math.sqrt(5000)
print("Box-Muller: mean %.3f sd %.3f KS D %.4f | crew: min %.3f s, mean %.3f s (theory %.3f s), KS D %.4f | critical %.4f"
      % (z.mean(), z.std(), D_z, c.min(), c.mean(), shift + math.exp(mu + sg**2 / 2), D_c, crit))
assert D_z < crit and D_c < crit and c.min() > shift
print("PASS 5.6")
```

### 5.9 Two-car discrete-event model: reduction and queueing checks
The team model (§4, *Two-car team model*) must (a) reproduce the single-car simulator **exactly** for both cars when interaction is switched off, over 300 random races; (b) leave the lead car untouched when the stops never overlap; (c) make the second car wait exactly $C+\tau-g_0$ when both cars stop on the same lap in a deterministic race (crew time $C$, turnaround $\tau$, grid gap $g_0$); and (d) never let a car wait when the stops are on different laps.

```python
sT = strategy([26], ["MEDIUM", "HARD"])
a_sc, b_sc = draw_team_scenarios(300, p, SEED + 9)
q_off = team_params(p, interact=False)
TA = np.array([simulate_team_race(sT, sT, q_off, a_sc, b_sc, i)["total_a"] for i in range(300)])
TB = np.array([simulate_team_race(sT, sT, q_off, a_sc, b_sc, i)["total_b"] for i in range(300)])
err_a = np.max(np.abs(TA - simulate_scenarios(sT, p, a_sc))); err_b = np.max(np.abs(TB - simulate_scenarios(sT, p, b_sc)))
print("(a) no interaction: max |team - single| car A %.1e s, car B %.1e s" % (err_a, err_b))
assert err_a < 1e-9 and err_b < 1e-9
pdet = linear_params(p); pdet.update(random=False, p_gv=0.0, p_gs=0.0, p_lap1_sc=0.0)
r = simulate_two_car_race(strategy([18], ["MEDIUM", "HARD"]), strategy([36], ["MEDIUM", "HARD"]), pdet, 0)
single = simulate_race(strategy([18], ["MEDIUM", "HARD"]), pdet, 0)["total_time"]
print("(b) non-overlapping stops: car A %.6f s vs single car %.6f s" % (r["total_a"], single))
assert abs(r["total_a"] - single) < 1e-6 and r["stack_delay_b"] == 0
r = simulate_two_car_race(sT, sT, pdet, 0)
tq = team_params(pdet)
expect = (tq["crew_shift_s"] + math.exp(tq["crew_mu"])) + tq["box_turn_s"] - tq["grid_gap_s"]
print("(c) same-lap stops: car B waited %.6f s, expected C + tau - g0 = %.6f s" % (r["stack_delay_b"], expect))
assert abs(r["stack_delay_b"] - expect) < 1e-9
R = run_team(strategy([20], ["MEDIUM", "HARD"]), strategy([32], ["MEDIUM", "HARD"]), p, 300, SEED + 10)
print("(d) different laps: max wait A %.2f s, B %.2f s over 300 random races" % (R["stack_delay_a"].max(), R["stack_delay_b"].max()))
assert R["stack_delay_a"].max() == 0 and R["stack_delay_b"].max() == 0
print("PASS 5.9")
```

### 6.3 How the answer changes with the SC onset probability $P_{GS}$
For each $P_{GS}$ in the sweep, 10,000 fresh common scenarios give paired differences (two-stop − one-stop, reactive − one-stop) with 95% CIs. The crossover is the $P_{GS}$ at which a difference changes sign (linear interpolation of the means). We also sweep $\alpha_{SC}$ at the baseline $P_{GS}$.

```python
sweep = []
for v in P_GS_SWEEP:
    q = with_params(p, p_gs=v)
    X = run_paired(SHORTLIST, q, 10_000, SEED + 100 + int(v * 1e4))
    sweep.append([v] + [mean_ci(X[:, j] - X[:, 0]) for j in (1, 2)])
def crossover(xs, ys):
    # first P_GS where the mean difference changes sign (linear interpolation)
    for x0, y0, x1, y1 in zip(xs, ys, xs[1:], ys[1:]):
        if y0 == 0:
            return x0
        if y0 * y1 < 0:
            return x0 + (x1 - x0) * y0 / (y0 - y1)
    return None
fig, ax = plt.subplots(figsize=(7, 4.2))
cross = {}
for j, lab in ((1, "two-stop − one-stop"), (2, "pit-if-neutralised − one-stop")):
    m = np.array([row[j][0] for row in sweep]); lo = np.array([row[j][1] for row in sweep]); hi = np.array([row[j][2] for row in sweep])
    ax.plot(P_GS_SWEEP, m, "o-", label=lab); ax.fill_between(P_GS_SWEEP, lo, hi, alpha=.25)
    cross[lab] = crossover(P_GS_SWEEP, m)
    print("%-32s" % lab, " ".join("%.3f:%+.2f[%+.2f,%+.2f]" % (v, a, b, c) for v, (a, b, c) in zip(P_GS_SWEEP, [row[j] for row in sweep])))
ax.axhline(0, color="k", lw=.8); ax.axvline(p["p_gs"], color="grey", ls=":", label="baseline $P_{GS}$")
ax.set_xlabel("$P_{GS}$ (per-lap Safety-Car onset probability)"); ax.set_ylabel("paired race-time difference [s]")
ax.set_title("Strategy difference vs SC onset probability (95% CI)"); ax.legend(); ax.grid(alpha=.3); fig.tight_layout()
fig.savefig(os.path.join(FIG_DIR, "results_pgs_sweep.png"), dpi=150); plt.show()
for k, v in cross.items():
    print("crossover (%s): %s" % (k, "P_GS ~ %.4f" % v if v is not None else "none in the swept range"))
```

### 6.5 Performance: plain loop vs vectorised NumPy
The same scenarios are run through the per-race Python loop and the vectorised simulator. The results must be identical, and the timings are the median of 3 runs.

**Why it is faster.** The loop advances one race at a time, so every lap costs dozens of Python-level operations (dictionary look-ups, a scalar RK4 step, branching on the track status). The vectorised version swaps the loops: it advances *all* races one lap at a time, so each RK4 stage, each lap-time formula and each pit decision is one NumPy array operation running in compiled code. The Python overhead is paid 62 times per strategy instead of $62N$ times, so the speed-up grows with $N$ until memory bandwidth dominates. Pre-drawing every random number per source (status, lap noise, crew) also keeps CRN exact: the two implementations read the same numbers and agree to $10^{-9}$ s.

```python
scen = draw_scenarios(2000, p, SEED + 7)
t_loop, a = timed(lambda: np.array([simulate_race_loop(best1, p, scen, i) for i in range(2000)]))
t_vec, b = timed(lambda: simulate_scenarios(best1, p, scen))
assert np.max(np.abs(a - b)) < 1e-9
print("loop %.3f s | vectorised %.4f s | speed-up %.1fx (identical results)" % (t_loop, t_vec, t_loop / t_vec))

# scaling with the number of replications (median of 3 runs each)
Ns_perf = [250, 500, 1000, 2000, 4000, 8000]
t_l, t_v = [], []
for n in Ns_perf:
    sc_ = draw_scenarios(n, p, SEED + 70 + n)
    t_l.append(timed(lambda: [simulate_race_loop(best1, p, sc_, i) for i in range(n)])[0] if n <= 4000 else np.nan)
    t_v.append(timed(lambda: simulate_scenarios(best1, p, sc_))[0])
t_l, t_v = np.array(t_l), np.array(t_v)
ok = np.isfinite(t_l)
fig, axs = plt.subplots(1, 2, figsize=(11, 4))
axs[0].loglog(np.array(Ns_perf)[ok], t_l[ok], "o-", label="plain Python loop (one race at a time)")
axs[0].loglog(Ns_perf, t_v, "s-", label="vectorised NumPy (all races per lap step)")
axs[0].set_xlabel("races simulated"); axs[0].set_ylabel("wall time [s]"); axs[0].set_title("Run time vs number of races")
axs[0].legend(); axs[0].grid(True, which="both", alpha=.3)
axs[1].semilogx(np.array(Ns_perf)[ok], t_l[ok] / t_v[ok], "o-", color="C2")
axs[1].set_xlabel("races simulated"); axs[1].set_ylabel("speed-up (loop / vectorised)"); axs[1].set_title("Measured speed-up")
axs[1].grid(True, which="both", alpha=.3); fig.tight_layout(); fig.savefig(os.path.join(FIG_DIR, "performance_scaling.png"), dpi=150); plt.show()
for n, a_, b_ in zip(Ns_perf, t_l, t_v):
    print("N = %5d  loop %s  vectorised %.4f s  -> %s races/s" % (n, "%.3f s" % a_ if a_ == a_ else "   (skipped)", b_, "{:,.0f}".format(n / b_)))
print("Stage-1 screening of %d strategies x 1,000 races would take ~%.0f s with the loop vs %.1f s vectorised"
      % (len(cands), len(cands) * t_l[2], len(cands) * t_v[2]))
```

### 6.6 Team extension: is double-stacking under a Safety Car worth it?
When a Safety Car comes out, both cars of a team want the cheap stop, but the second car must queue behind the first in the single pit box. Three team policies are compared on the same 4,000 race scenarios (CRN across policies), using the best reactive one-stop from Stage 1:
* **both react**: both cars stop at the first SC/VSC in the window, so they double-stack whenever it happens (otherwise A stops on lap $L$ and B on $L+1$);
* **split**: car A reacts, car B keeps its fixed one-stop plan (staggered one lap after A's planned lap, as teams do to avoid a planned double-stack);
* **both fixed**: neither car reacts (A on lap $L$, B on lap $L+1$).

The output is the team's total race time (A + B) relative to *both fixed*, with 95% CIs, and car B's queueing delay. The sweep over $P_{GS}$ shows when a team should split its cars.

```python
react_ = bestR
fixed_ = strategy(react_["pit_laps"], react_["compounds"])
fixed_b = strategy([react_["pit_laps"][0] + 1], react_["compounds"])     # B staggered one lap to avoid a planned stack
react_b = strategy([react_["pit_laps"][0] + 1], react_["compounds"], reactive_window=(react_["reactive_window"][0], react_["pit_laps"][0] + 1))
policies = {"both react": (react_, react_b), "split (A reacts)": (react_, fixed_b), "both fixed": (fixed_, fixed_b)}
team_rows, team_sweep = {}, {k: [] for k in policies}
PGS_TEAM = [0.0, 0.02, 0.04, 0.06]
for v in PGS_TEAM:
    q = with_params(p, p_gs=v)
    res = {k: run_team(sa, sb, q, 4000, SEED + 600 + int(v * 1e4)) for k, (sa, sb) in policies.items()}
    base = res["both fixed"]["total_a"] + res["both fixed"]["total_b"]
    for k, R in res.items():
        team_sweep[k].append(mean_ci(R["total_a"] + R["total_b"] - base))
    if abs(v - p["p_gs"]) < 1e-12:
        team_rows = res
print("Baseline P_GS = %.3f, 4,000 common race scenarios per policy" % p["p_gs"])
for k, R in team_rows.items():
    w = R["stack_delay_b"]
    print("  %-18s car B waited in %.1f%% of races, mean wait when it did %.2f s; traffic %.2f s" %
          (k, 100 * (w > 0).mean(), w[w > 0].mean() if (w > 0).any() else 0.0, R["traffic_b"].mean()))
fig, axs = plt.subplots(1, 2, figsize=(11, 4))
for k in ("both react", "split (A reacts)"):
    m = np.array([x[0] for x in team_sweep[k]]); lo = np.array([x[1] for x in team_sweep[k]]); hi = np.array([x[2] for x in team_sweep[k]])
    axs[0].plot(PGS_TEAM, m, "o-", label=k); axs[0].fill_between(PGS_TEAM, lo, hi, alpha=.25)
    print("%-18s team gain vs both fixed: %s" % (k, "  ".join("%.2f:%+.2f[%+.2f,%+.2f]" % (v, *x) for v, x in zip(PGS_TEAM, team_sweep[k]))))
R_r, R_s = team_rows["both react"], team_rows["split (A reacts)"]
dd = mean_ci((R_r["total_a"] + R_r["total_b"]) - (R_s["total_a"] + R_s["total_b"]))
print("Double-stack vs split, team time (baseline): %+.2f s [%+.2f, %+.2f]" % dd)
axs[0].axhline(0, color="k", lw=.8); axs[0].set_xlabel("$P_{GS}$"); axs[0].set_ylabel("team time vs both fixed [s]")
axs[0].set_title("Team policy under Safety-Car risk (95% CI)"); axs[0].legend(); axs[0].grid(alpha=.3)
w = team_rows["both react"]["stack_delay_b"]
axs[1].hist(w[w > 0], bins=40, color="C1", alpha=.8)
axs[1].set_xlabel("car B queueing delay when double-stacked [s]"); axs[1].set_ylabel("races"); axs[1].set_title("Double-stack delay (both react, baseline)")
axs[1].grid(alpha=.3); fig.tight_layout(); fig.savefig(os.path.join(FIG_DIR, "team_double_stack.png"), dpi=150); plt.show()
```

### 6.7 Real-data grounding: calibration and held-out validation (Singapore 2008–2025)
The data are 16 Singapore races from Jolpica-F1 (lap times and pit stops) plus 2023–2025 stints, compounds, stationary times and race-control messages from OpenF1 (`data/`, CC BY-NC-SA 4.0). The calibration code (`src/calibrate.py`) is shown in the next cell. The public lap data has no status flag, so Green / VSC / SC is **inferred from the field's lap times**: a neutralised lap is more than 18% slower than Green pace, and a neutralised period is a Safety Car when the field closes up behind it (gaps < 1.5 s), otherwise a VSC. The classifier is checked against OpenF1's race-control messages.

Then: $\hat P$ by maximum likelihood from transition counts, a Beta posterior for $P_{GS}$, the phase-type SC duration law, the neutralised pace ratios $k_S,k_V$, the 2025 (80 km/h) Green pit loss and stationary-time law, the measured SC pit-loss discount $\alpha_{SC}$, and per-compound wear rates from 2023–2025 stints.

**Held-out validation:** wear rates fitted on 2023–2024 predict the 2025 stints; the SC duration law fitted on 2008–2017 is tested on 2018–2025 (KS, parametric bootstrap); and $P_{GS}$ from 2008–2017 predicts the number of SC onsets in 2018–2025.

```python
# src/calibrate.py (real-data calibration)
import csv, io, urllib.request


GREEN, VSC, SC = 0, 1, 2
RAW_URL = "https://raw.githubusercontent.com/dinukacodes/f1-pit-strategy-simulator/main/data/"
SOURCE = ("Jolpica-F1 (Ergast successor) https://github.com/jolpica/jolpica-f1 and OpenF1 https://openf1.org, "
          "Singapore Grand Prix 2008-2025, CC BY-NC-SA 4.0")
NEUTRAL_RATIO = 1.18     # lap slower than 118% of Green pace -> neutralised (lap >= 2)
LAP1_RATIO = 1.25        # a standing start is slow anyway, so lap 1 needs a higher bar
MIN_STINTS = 30         # a compound needs >= 30 clean stints for its own wear estimate
SC_GAP_S = 1.5           # field closed up to < 1.5 s median gap -> Safety Car (bunching)


# ---------------------------------------------------------------------------
# Loading
# ---------------------------------------------------------------------------

def _find_data_dir():
    here = os.path.dirname(os.path.abspath(globals().get("__file__", "x")))
    for d in ("data", "../data", os.path.join(here, "..", "data")):
        if os.path.exists(os.path.join(d, "singapore_laps.csv")):
            return d
    return None


def read_csv(name, data_dir=None):
    """Read data/<name> as a dict of numpy arrays (numbers converted where possible).

    Falls back to the copy on GitHub when the file is not available locally (Colab)."""
    d = data_dir or _find_data_dir()
    if d is not None and os.path.exists(os.path.join(d, name)):
        text = open(os.path.join(d, name), encoding="utf-8").read()
    else:
        with urllib.request.urlopen(RAW_URL + name, timeout=60) as r:
            text = r.read().decode("utf-8")
    rows = list(csv.DictReader(io.StringIO(text)))
    out = {}
    for k in rows[0]:
        col = [r[k] for r in rows]
        try:
            out[k] = np.array([float(v) if v != "" else np.nan for v in col])
        except ValueError:
            out[k] = np.array(col)
    return out


def load_all(data_dir=None):
    names = ["singapore_laps", "singapore_pit_stops", "singapore_races", "openf1_stints",
             "openf1_laps", "openf1_pit", "openf1_race_control"]
    return {n: read_csv(n + ".csv", data_dir) for n in names}


# ---------------------------------------------------------------------------
# Track status from lap times
# ---------------------------------------------------------------------------

def _runs(mask):
    """[(start, end)] index runs where mask is True (inclusive)."""
    out, start = [], None
    for i, v in enumerate(list(mask) + [False]):
        if v and start is None:
            start = i
        elif not v and start is not None:
            out.append((start, i - 1))
            start = None
    return out


def race_matrix(laps, year):
    """(n_laps, n_drivers) lap-time matrix [s] for one race, NaN where a car did not run."""
    m = laps["year"] == year
    drivers = sorted(set(laps["driver"][m]))
    n = int(laps["lap"][m].max())
    T = np.full((n, len(drivers)), np.nan)
    col = {d: j for j, d in enumerate(drivers)}
    for l, d, t in zip(laps["lap"][m], laps["driver"][m], laps["time_s"][m]):
        T[int(l) - 1, col[d]] = t
    return T, drivers


def infer_status(laps, year):
    """Infer Green / VSC / SC for every lap of one race.

    A lap is neutralised when the field's median lap time exceeds NEUTRAL_RATIO
    times the race's Green pace. A neutralised period is a Safety Car when the
    field has closed up behind it (median gap among the top ten < SC_GAP_S at
    its end), otherwise a VSC (which keeps the gaps). Returns dict(status,
    ratio, ref_s, runs) with runs = [(first_lap, last_lap, kind, end_gap_s)].
    """
    T, _ = race_matrix(laps, year)
    med = np.nanmedian(T, axis=1)
    ref = np.median(med[1:])
    ratio = med / ref
    neutral = ratio > NEUTRAL_RATIO
    neutral[0] = ratio[0] > LAP1_RATIO
    cum = np.cumsum(np.nan_to_num(T, nan=0.0), axis=0)
    cum[~np.isfinite(T)] = np.nan
    status = np.zeros(len(med), dtype=int)
    runs = []
    for a, b in _runs(neutral):
        row = np.sort(cum[b][np.isfinite(cum[b])])[:10]
        end_gap = float(np.median(np.diff(row))) if len(row) > 2 else np.nan
        if year < 2015:
            kind = SC                                   # the VSC exists from 2015
        elif b == a:
            kind = VSC                                  # an SC takes at least two laps (deploy + in)
        else:
            kind = SC if end_gap < SC_GAP_S else VSC
        status[a:b + 1] = kind
        runs.append((a + 1, b + 1, kind, end_gap))
    return dict(status=status, ratio=ratio, ref_s=float(ref), runs=runs)


def all_status(data):
    years = [int(y) for y in data["singapore_races"]["year"]]
    return {y: infer_status(data["singapore_laps"], y) for y in years}


def race_control_check(data, status):
    """OpenF1 race-control Safety-Car / VSC messages next to the inferred status of that lap."""
    rc = data["openf1_race_control"]
    out = []
    for y, l, msg in zip(rc["year"], rc["lap"], rc["message"]):
        y, l = int(y), int(l)
        if "SAFETY CAR" in msg and "INFRINGEMENT" not in msg and y in status:
            out.append((y, l, msg, int(status[y]["status"][l - 1])))
    return out


# ---------------------------------------------------------------------------
# Markov chain estimation
# ---------------------------------------------------------------------------

def transition_counts(seqs):
    """3x3 lap-to-lap transition counts n_ij over all races."""
    C = np.zeros((3, 3), dtype=int)
    for s in seqs:
        for a, b in zip(s[:-1], s[1:]):
            C[a, b] += 1
    return C


def mle_matrix(C):
    """Maximum-likelihood transition matrix P_ij = n_ij / sum_k n_ik."""
    C = np.asarray(C, float)
    return C / np.maximum(C.sum(1, keepdims=True), 1)


def durations(seqs, state):
    out = []
    for s in seqs:
        for a, b in _runs(np.asarray(s) == state):
            out.append(b - a + 1)
    return np.array(out)


def onset_counts(seqs, state):
    """Green -> state onsets (laps >= 2) and the number of Green origins."""
    n_on, n_g = 0, 0
    for s in seqs:
        for l in range(1, len(s)):
            if s[l - 1] == GREEN:
                n_g += 1
                n_on += int(s[l] == state)
    return n_on, n_g


def beta_ci(k, n, level=0.95, prior=(1.0, 1.0), seed=0):
    """Posterior mean and equal-tailed credible interval for a Bernoulli probability."""
    rng = np.random.default_rng(seed)
    a, b = prior[0] + k, prior[1] + n - k
    d = rng.beta(a, b, 400_000)
    return a / (a + b), float(np.quantile(d, (1 - level) / 2)), float(np.quantile(d, (1 + level) / 2))


def phase_type_fit(D):
    """m = min duration, p_SS = sum(D - m) / sum(D - m + 1) (maximum likelihood)."""
    D = np.asarray(D, float)
    m = int(D.min())
    return m, float((D - m).sum() / (D - m + 1).sum())


def phase_type_cdf(x, m, pss):
    x = np.floor(np.asarray(x, float))
    return np.where(x < m, 0.0, 1.0 - pss ** (x - m + 1))


def ks_discrete(D, m, pss):
    """KS distance between the empirical law of D and the phase-type law."""
    D = np.sort(np.asarray(D, float))
    xs = np.arange(1, D.max() + 2)
    emp = np.array([(D <= x).mean() for x in xs])
    return float(np.max(np.abs(emp - phase_type_cdf(xs, m, pss))))


def ks_pvalue_bootstrap(D, m, pss, n_boot=4000, seed=1):
    """Parametric-bootstrap p-value of the KS distance for a fixed (held-out) law."""
    rng = np.random.default_rng(seed)
    d0 = ks_discrete(D, m, pss)
    sims = m + rng.geometric(1 - pss, size=(n_boot, len(D))) - 1 if pss > 0 else np.full((n_boot, len(D)), m)
    ds = np.array([ks_discrete(s, m, pss) for s in sims])
    return d0, float((ds >= d0 - 1e-12).mean())


def poisson_two_sided_p(k, lam):
    pmf = lambda j: math.exp(-lam + j * math.log(lam) - math.lgamma(j + 1))
    pk = pmf(k)
    return min(1.0, sum(pmf(j) for j in range(0, int(lam * 4 + 30)) if pmf(j) <= pk * (1 + 1e-9)))


# ---------------------------------------------------------------------------
# Neutralised pace and pit loss
# ---------------------------------------------------------------------------

def neutral_ratios(status):
    """Median lap-time ratio on the interior laps of SC and VSC periods (edge laps are partial)."""
    out = {}
    for kind in (SC, VSC):
        r = []
        for y, s in status.items():
            for a, b, k, _ in s["runs"]:
                if k == kind:
                    inner = range(a + 1, b) if b - a >= 2 else range(a, b + 1)
                    r += [s["ratio"][l - 1] for l in inner if l > 1]
        out[kind] = (float(np.median(r)), len(r)) if r else (np.nan, 0)
    return out


def stop_losses(data, status, year):
    """Net pit losses [s] of one race: in-lap + out-lap minus two reference laps.

    Green stops use the driver's median clean Green lap; stops whose in-lap and
    out-lap are both under SC use the field's median on those laps (the SC pace)."""
    T, drivers = race_matrix(data["singapore_laps"], year)
    col = {d: j for j, d in enumerate(drivers)}
    ps = data["singapore_pit_stops"]
    m = ps["year"] == year
    st = status[year]["status"]
    pit_laps = {}
    for d, l in zip(ps["driver"][m], ps["lap"][m]):
        pit_laps.setdefault(d, set()).update({int(l), int(l) + 1})
    med = np.nanmedian(T, axis=1)
    green, under_sc = [], []
    for d, l in zip(ps["driver"][m], ps["lap"][m]):
        l, j = int(l), col.get(d)
        if j is None or l + 1 > T.shape[0] or l < 2:
            continue
        tin, tout = T[l - 1, j], T[l, j]
        if not (np.isfinite(tin) and np.isfinite(tout)):
            continue
        if st[l - 1] == GREEN and st[l] == GREEN:
            clean = [T[k, j] for k in range(1, T.shape[0])
                     if st[k] == GREEN and (k + 1) not in pit_laps.get(d, ()) and np.isfinite(T[k, j])]
            if len(clean) >= 10:
                loss = tin + tout - 2 * np.median(clean)
                if 0 < loss < 60:
                    green.append(loss)
        elif st[l - 1] == SC and st[l] == SC:
            loss = tin + tout - med[l - 1] - med[l]
            if -10 < loss < 60:
                under_sc.append(loss)
    return dict(green=np.array(green), sc=np.array(under_sc))


def fit_shifted_lognormal(x, shift):
    y = np.log(np.asarray(x) - shift)
    return float(y.mean()), float(y.std(ddof=1))


# ---------------------------------------------------------------------------
# Tyre wear from stints
# ---------------------------------------------------------------------------

def clean_stint_laps(data, status, years):
    """Clean Green laps with compound and tyre age: list of (year, stint_key, compound, age, lap, t)."""
    S, L = data["openf1_stints"], data["openf1_laps"]
    lap_t = {(int(y), int(dn), int(l)): (t, int(po)) for y, dn, l, t, po in
             zip(L["year"], L["driver_number"], L["lap"], L["lap_s"], L["pit_out_lap"])}
    rows = []
    for y, dn, sn, c, a, b, age0 in zip(S["year"], S["driver_number"], S["stint"], S["compound"],
                                        S["lap_start"], S["lap_end"], S["tyre_age_at_start"]):
        y, dn, a, b = int(y), int(dn), int(a), int(b)
        if y not in years or c not in ("SOFT", "MEDIUM", "HARD") or y not in status:
            continue
        st = status[y]["status"]
        pts = []
        for l in range(a + 1, b):                       # drop the out-lap (a) and the in-lap (b)
            t, po = lap_t.get((y, dn, l), (np.nan, 1))
            if l >= 3 and np.isfinite(t) and not po and l <= len(st) and st[l - 1] == GREEN \
                    and st[l - 2] == GREEN:             # also drop the restart lap after a neutralisation
                pts.append((l, (age0 if age0 == age0 else 0) + (l - a), t))
        if len(pts) >= 6:
            ts = np.array([q[2] for q in pts])
            keep = ts < 1.03 * np.median(ts)            # traffic, mistakes, blue flags
            rows += [(y, "%d-%d-%d" % (y, dn, int(sn)), c, age, l, t) for (l, age, t), k in zip(pts, keep) if k]
    return rows


def fit_wear(rows, fuel_slope_s_per_lap):
    """Per-compound wear slope [s per lap of tyre age] with stint fixed effects.

    Lap times are corrected for fuel burn (t + fuel_slope * (lap - 1)), demeaned
    within each stint, and the slope is pooled least squares on the demeaned data.
    Returns {compound: (slope, standard error, n_laps, n_stints)}."""
    out = {}
    for c in ("SOFT", "MEDIUM", "HARD"):
        r = [x for x in rows if x[2] == c]
        keys = sorted(set(x[1] for x in r))
        if len(keys) < 2:
            continue
        xs, ys = [], []
        for k in keys:
            rr = [x for x in r if x[1] == k]
            a = np.array([x[3] for x in rr], float)
            t = np.array([x[5] + fuel_slope_s_per_lap * (x[4] - 1) for x in rr])
            xs.append(a - a.mean())
            ys.append(t - t.mean())
        x, y = np.concatenate(xs), np.concatenate(ys)
        b = float((x * y).sum() / (x * x).sum())
        resid = y - b * x
        dof = max(len(x) - len(keys) - 1, 1)
        out[c] = (b, float(math.sqrt((resid ** 2).sum() / dof / (x * x).sum())), len(x), len(keys))
    return out


def stint_prediction_rmse(rows, slopes, fuel_slope_s_per_lap, n_anchor=3):
    """Held-out test: anchor each stint on its first n_anchor clean laps and predict the rest.

    Returns (rmse_model, rmse_no_wear, n_predicted_laps)."""
    e_m, e_f = [], []
    for k in sorted(set(x[1] for x in rows)):
        rr = sorted([x for x in rows if x[1] == k], key=lambda x: x[4])
        if len(rr) <= n_anchor + 2 or rr[0][2] not in slopes:
            continue
        a = np.array([x[3] for x in rr], float)
        t = np.array([x[5] + fuel_slope_s_per_lap * (x[4] - 1) for x in rr])
        b = slopes[rr[0][2]]
        c0 = np.mean(t[:n_anchor] - b * a[:n_anchor])
        e_m += list(t[n_anchor:] - (c0 + b * a[n_anchor:]))
        e_f += list(t[n_anchor:] - np.mean(t[:n_anchor]))
    e_m, e_f = np.array(e_m), np.array(e_f)
    return float(np.sqrt(np.mean(e_m ** 2))), float(np.sqrt(np.mean(e_f ** 2))), len(e_m)


# ---------------------------------------------------------------------------
# Everything together
# ---------------------------------------------------------------------------

def calibrate(data=None, p=None):
    """Run the full calibration.

    Returns a dict with the inferred status, the Markov-chain estimates, the pit
    and tyre fits, 'params' (changes to apply with f1sim.with_params) and
    'validation' (held-out metrics)."""
    data = data or load_all()
    p = p or default_params()
    status = all_status(data)
    years = sorted(status)
    seqs = [status[y]["status"] for y in years]
    C = transition_counts(seqs)
    P = mle_matrix(C)
    lap1_sc = sum(int(s[0] == SC) for s in seqs)
    n_gs, n_g = onset_counts(seqs, SC)
    vsc_seqs = [status[y]["status"] for y in years if y >= 2015]
    n_gv, n_gv_g = onset_counts(vsc_seqs, VSC)
    D_sc = durations(seqs, SC)
    D_v = durations(vsc_seqs, VSC)
    m_sc, pss = phase_type_fit(D_sc)
    ratios = neutral_ratios(status)
    k_sc = ratios[SC][0]

    # crew time and Green pit loss (2025, 80 km/h pit lane)
    of = data["openf1_pit"]
    stop25 = of["stop_s"][(of["year"] == 2025) & np.isfinite(of["stop_s"])]
    shift = float(np.floor((stop25.min() - 0.1) * 10) / 10)
    mu, sg = fit_shifted_lognormal(stop25, shift)
    L25 = stop_losses(data, status, 2025)
    dT_G = float(np.median(L25["green"]) - np.median(stop25))
    # SC discount on the pit loss, measured: in a race with both Green and SC stops the
    # crew time cancels in median(Green loss) - median(SC loss) = alpha_SC * dT_G(year),
    # with dT_G(year) = median(Green loss) - E[C]. Geometry (R10) predicts
    # alpha_SC = (k_SC - 1) T_bypass / dT_G; the implied T_bypass is reported too.
    mean_crew = shift + math.exp(mu + sg ** 2 / 2)
    tb = []
    for y in years:
        if y >= 2011:
            Ly = stop_losses(data, status, y)
            if len(Ly["sc"]) >= 3 and len(Ly["green"]) >= 3:
                g, s_ = np.median(Ly["green"]), np.median(Ly["sc"])
                tb.append(dict(year=y, n_sc=len(Ly["sc"]), n_green=len(Ly["green"]), green_s=float(g), sc_s=float(s_),
                               alpha_sc=float((g - s_) / (g - mean_crew))))
    alpha_sc = float(np.median([x["alpha_sc"] for x in tb])) if tb else p["alpha_sc"]
    t_bypass = alpha_sc * dT_G / (k_sc - 1)

    # tyre wear: fit on 2023-2024, test on 2025; the final fit uses all three seasons
    fuel_slope = p["gamma_s_per_kg"] * p["burn_kg_per_lap"]
    tr = clean_stint_laps(data, status, {2023, 2024})
    te = clean_stint_laps(data, status, {2025})
    allr = clean_stint_laps(data, status, {2023, 2024, 2025})
    w_tr, w_all = fit_wear(tr, fuel_slope), fit_wear(allr, fuel_slope)
    rmse_m, rmse_f, n_te = stint_prediction_rmse(te, {c: v[0] for c, v in w_tr.items()}, fuel_slope)
    # the model's average wear growth per lap is ~ k_c * (1 + beta * m_mid) (eta is small)
    m_mid = p["m0_kg"] - p["burn_kg_per_lap"] * (p["n_laps"] / 2)
    k_fit = dict(p["k_s_per_lap"])
    for c, v in w_all.items():
        if v[3] >= MIN_STINTS:
            k_fit[c] = max(v[0], 0.005) / (1 + p["beta_per_kg"] * m_mid)
    for c in ("SOFT", "MEDIUM", "HARD"):            # thinly observed compounds keep the assumed ratio to MEDIUM
        if c not in w_all or w_all[c][3] < MIN_STINTS:
            k_fit[c] = k_fit["MEDIUM"] * p["k_s_per_lap"][c] / p["k_s_per_lap"]["MEDIUM"]

    # held-out Markov-chain tests: fit on 2008-2017, test on 2018-2025
    old = [status[y]["status"] for y in years if y <= 2017]
    new = [status[y]["status"] for y in years if y >= 2018]
    m_old, pss_old = phase_type_fit(durations(old, SC))
    D_new = durations(new, SC)
    ks_d, ks_p = ks_pvalue_bootstrap(D_new, m_old, pss_old)
    k_old, n_old = onset_counts(old, SC)
    k_new, n_new = onset_counts(new, SC)
    lam = k_old / n_old * n_new

    params = dict(
        p_gs=n_gs / n_g, p_gv=n_gv / n_gv_g if n_gv else p["p_gv"],
        p_vv=float(P[VSC, VSC]) if C[VSC].sum() else p["p_vv"],
        p_lap1_sc=lap1_sc / len(seqs), sc_min_laps=m_sc, p_ss=pss,
        k_sc=k_sc, k_vsc=ratios[VSC][0] if ratios[VSC][1] else p["k_vsc"],
        crew_shift_s=shift, crew_mu=mu, crew_sigma=sg,
        t_bypass_s=t_bypass, t_pit_s=dT_G + t_bypass, k_s_per_lap=k_fit,
    )
    vsc_geom = (params["k_vsc"] - 1) * t_bypass / dT_G
    params.update(alpha_sc=alpha_sc, alpha_vsc=min(vsc_geom, 0.95), dT_green_s=dT_G)
    return dict(
        status=status, years=years, transition_counts=C, P=P, n_races=len(seqs), source=SOURCE,
        lap1_sc=lap1_sc, onsets_sc=(n_gs, n_g), onsets_vsc=(n_gv, n_gv_g), p_gs_posterior=beta_ci(n_gs, n_g),
        sc_durations=D_sc, vsc_durations=D_v, sc_fit=(m_sc, pss), ratios=ratios,
        crew=dict(stops=stop25, shift=shift, mu=mu, sigma=sg, mean=shift + math.exp(mu + sg ** 2 / 2)),
        pit2025=L25, dT_G=dT_G, sc_stop_years=tb, alpha_sc=alpha_sc, t_bypass=t_bypass, wear_train=w_tr, wear_all=w_all, wear_rows=allr,
        race_control=race_control_check(data, status), params=params,
        validation=dict(
            heldout_wear_rmse_s=rmse_m, heldout_no_wear_rmse_s=rmse_f, heldout_wear_laps=n_te,
            heldout_sc_duration_ks=ks_d, heldout_sc_duration_ks_p=ks_p,
            heldout_sc_onsets_observed=k_new, heldout_sc_onsets_expected=lam,
            heldout_sc_onsets_poisson_p=poisson_two_sided_p(k_new, lam),
            train_sc_fit=(m_old, pss_old), test_sc_durations=D_new,
        ),
    )
```

### 6.8 The answer with calibrated parameters, and parameter uncertainty
The calibrated values replace the assumed ones (compound pace offsets, fuel and noise stay as before). Stage 1 and Stage 2 are re-run on fresh seeds, so the shortlist is chosen again in the calibrated world.

Then **parameter uncertainty is propagated** through the decision (X5): 300 parameter sets are drawn ($P_{GS}$ from its Beta posterior, each wear rate from $N(\hat k_c, \widehat{se}^2)$, $\alpha_{SC}$ resampled from the per-race estimates), and each set is evaluated on 2,000 paired races. The result is the probability that each plan is best once both race randomness and parameter uncertainty are counted.

```python
pc = with_params(p, **CAL["params"])
cands_c = candidate_strategies(pc) + [strategy([L], c, reactive_window=(12, L)) for c in (("MEDIUM", "HARD"), ("HARD", "MEDIUM"), ("SOFT", "HARD")) for L in range(14, 46, 2)]
ranked_c = screen(cands_c, pc, 1000, SEED + 211)
pick = lambda pred: next(s for m, s in ranked_c if pred(s))
SL_C = [pick(lambda s: len(s["pit_laps"]) == 1 and s["reactive_window"] is None), pick(lambda s: len(s["pit_laps"]) == 2),
        pick(lambda s: s["reactive_window"] is not None)]
rows_c, Xc = paired_table(SL_C, pc, 10_000, SEED + 222, ref=0, z=bonferroni_z(2))
print("Calibrated world (alpha_SC = %.2f, alpha_VSC = %.2f, dT_G = %.1f s, P_GS = %.4f):" % (pc["alpha_sc"], pc["alpha_vsc"], pc["dT_green_s"], pc["p_gs"]))
for r_ in rows_c:
    print("  %-34s mean %.2f s   diff vs one-stop %+6.2f s [%+.2f, %+.2f]" % (r_["name"], r_["mean_s"], r_["diff_s"], r_["diff_low_s"], r_["diff_high_s"]))

rng_u = np.random.default_rng(SEED + 300)
k_gs, n_gs = CAL["onsets_sc"]
alphas_obs = np.array([r_["alpha_sc"] for r_ in CAL["sc_stop_years"]])
U = []
for d in range(300):
    ks = dict(pc["k_s_per_lap"])
    m_mid_ = p["m0_kg"] - p["burn_kg_per_lap"] * p["n_laps"] / 2
    for c_, (b_, se_, _, ns_) in CAL["wear_all"].items():
        if ns_ >= MIN_STINTS:
            ks[c_] = max(1e-3, rng_u.normal(b_, se_)) / (1 + p["beta_per_kg"] * m_mid_)
    ks["SOFT"] = ks["MEDIUM"] * p["k_s_per_lap"]["SOFT"] / p["k_s_per_lap"]["MEDIUM"]
    q = with_params(pc, p_gs=float(rng_u.beta(1 + k_gs, 1 + n_gs - k_gs)), k_s_per_lap=ks,
                    alpha_sc=float(np.clip(np.median(rng_u.choice(alphas_obs, len(alphas_obs))), 0, 1)), alpha_vsc=pc["alpha_vsc"])
    X = run_paired(SL_C, q, 2000, SEED + 5000 + d)
    U.append(X.mean(0))
U = np.array(U)
best = np.bincount(U.argmin(1), minlength=3) / len(U)
for j, s in enumerate(SL_C):
    print("P(best) %-34s %.2f" % (s["name"], best[j]))
d1, d2 = U[:, 1] - U[:, 0], U[:, 2] - U[:, 0]
print("two-stop - one-stop: median %+.2f s, 90%% interval [%+.2f, %+.2f]" % (np.median(d1), *np.quantile(d1, [.05, .95])))
print("reactive - one-stop: median %+.2f s, 90%% interval [%+.2f, %+.2f]" % (np.median(d2), *np.quantile(d2, [.05, .95])))
fig, ax = plt.subplots(figsize=(7, 4))
ax.hist(d1, bins=30, alpha=.6, label="two-stop − one-stop"); ax.hist(d2, bins=30, alpha=.6, label="pit-if-neutralised − one-stop")
ax.axvline(0, color="k", lw=.8); ax.set_xlabel("expected paired difference [s] for one parameter draw"); ax.set_ylabel("parameter draws")
ax.set_title("Decision under parameter uncertainty (300 draws × 2,000 races)"); ax.legend(); ax.grid(alpha=.3)
fig.tight_layout(); fig.savefig(os.path.join(FIG_DIR, "uncertainty_propagation.png"), dpi=150); plt.show()

# where does the calibrated answer switch?
sw_c = []
for v in [0.0, 0.0025, 0.005, 0.01, 0.015, 0.02, 0.03, 0.04]:
    X = run_paired(SL_C, with_params(pc, p_gs=v, alpha_sc=pc["alpha_sc"], alpha_vsc=pc["alpha_vsc"]), 10_000, SEED + 900 + int(v * 1e4))
    sw_c.append((v, mean_ci(X[:, 2] - X[:, 0]), mean_ci(X[:, 1] - X[:, 0])))
for v, r_, t_ in sw_c:
    print("calibrated P_GS = %.4f: reactive - one-stop %+.2f [%+.2f, %+.2f] s | two-stop - one-stop %+.2f [%+.2f, %+.2f] s" % (v, *r_, *t_))
cross_c = crossover([x[0] for x in sw_c], [x[1][0] for x in sw_c])
print("calibrated switch point (reactive vs one-stop): %s" % ("P_GS ~ %.4f" % cross_c if cross_c is not None else "none in range"))
```

**Team extension (§6.6).** When both cars stop under the same neutralisation, the second car loses about 3 s queueing in the box (crew time plus turnaround minus the gap). The SC discount on its stop is much larger, so double-stacking still beats leaving car B out, by the margin printed in §6.6. Both reactive team policies beat the fixed plans, and the advantage grows with $P_{GS}$.

**Real-data findings (§6.7–6.8).** The data confirm the baseline Safety-Car rate ($\hat P_{GS}\approx0.022$, 95% interval roughly 0.014–0.034, against an assumed 0.020) and the minimum SC length (2 laps, mean ≈ 3.7 laps), and the duration law and onset rate both pass their held-out tests on 2018–2025. Two inputs change materially. First, SC laps at Singapore run at about **1.38×** Green pace, not the 1.60× of the literature. Second, the **measured pit-loss discount under a Safety Car is about 0.86**, twice the 0.43 that the uniform-slowdown geometry predicts. A stop under SC costs only a few seconds because the pit-lane entry and exit are close to the SC queue. With that discount, *pit-if-neutralised* becomes the best plan in the calibrated world, about 4 s faster than the best one-stop with a CI far from zero, and it remains best in every one of 300 parameter draws. In the calibrated sweep the reactive plan wins across the whole range, even at $P_{GS}=0$, because VSC periods and lap-1 Safety Cars still occur. So at Singapore the question is no longer *whether* to react but how wide the window should be.

**Interpretation (baseline run, master seed 20251005; re-check the printed numbers if you change parameters).**
* **Fixed windows gain little from safety cars.** A planned stop falls under SC only with probability $p_S(L)$, a few percent per lap, so the expected discount per stop is small (§3.6). The stage-2 table shows the best one-stop ahead of the best two-stop by several seconds, with a CI far from zero.
* **More safety cars make the two-stop *worse*, not better.** SC laps cheapen stops (via $\alpha_{SC}$), but they also replace Green laps, where degradation costs time, and they slow tyre wear ($\varphi_S$). Fresh tyres are therefore worth less, and in this model the second effect dominates. So the two-stop never overtakes the one-stop in the swept range.
* **The value of an SC comes from reacting to it.** The pit-if-neutralised strategy's gap to the one-stop shrinks steadily as $P_{GS}$ grows. It crosses zero at the printed switch point, about twice the baseline onset rate.
* **What drives the answer:** the HARD wear rate and $P_{GS}$ dominate the tornado chart, while $\alpha_{SC}$ and the crew-time spread barely matter. Tyre wear and the SC discount are therefore the inputs that §6.7 calibrates from real data.
* **CRN is extremely effective here:** under H6, SC laps are identical for all strategies, so they cancel. The measured variance ratio is in the thousands, far beyond the ≥ 2 a generic model would give.

**Limitations.**
1. Single car: no track position, overtaking or traffic. At Singapore, track position matters a lot, and this model ignores it.
2. The objective is expected race time, not finishing position or points.
3. Track status is *inferred* from lap times (public lap data has no status flag). It matches OpenF1's race-control messages for 2023, but short VSC periods before 2015 cannot exist and edge laps are partial. Compound pace offsets, the fuel effect and the SOFT wear rate (too few soft stints) are still assumed.
4. A time-homogeneous incident rate after lap 1, and no VSC→SC escalation.
5. Pit losses assume the 2025 80 km/h pit-lane limit.
6. No weather, red flags, or slow-stop / unsafe-release events.
7. Sixteen races give about 19 SC onsets, so $P_{GS}$ is uncertain. This is why the Beta posterior is propagated through the decision (§6.8).
8. The team model has two cars and no other traffic, and the no-passing rule applies only between team-mates.

**Future work:** a Bellman (dynamic-programming) optimal reactive policy over (lap, status, tyre age, stops left); a full-field traffic model built on the team DES; per-team crew-time distributions; and an in-browser Pyodide build of the Python model.
